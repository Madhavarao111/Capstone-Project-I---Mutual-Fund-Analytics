## Live NAV Data Collection

This section fetches the latest NAV history from the mfapi.in API for selected mutual fund schemes and stores the data as CSV files.

```python
import requests
import os

# Create folder if it doesn't exist
live_nav_path = BASE_DIR / "data" / "raw" / "live_nav"
live_nav_path.mkdir(parents=True, exist_ok=True)
```

```python
schemes = {
    "HDFC_Top100":125497,
    "SBI_Bluechip":119551,
    "ICICI_Bluechip":120503,
    "Nippon_Large_Cap":118632,
    "Axis_Bluechip":119092,
    "Kotak_Bluechip":120841
}

schemes
```

```python
for fund_name, scheme_code in schemes.items():

    url = f"https://api.mfapi.in/mf/{scheme_code}"

    response = requests.get(url)

    if response.status_code == 200:

        data = response.json()

        nav_df = pd.DataFrame(data["data"])

        file_name = live_nav_path / f"{fund_name}.csv"

        nav_df.to_csv(file_name, index=False)

        print(f"Saved: {file_name}")

    else:
        print(f"Failed: {fund_name}")
```
